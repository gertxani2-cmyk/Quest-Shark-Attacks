```python
def extract_month(val):
    if pd.isna(val):
        return None
    val_str = str(val).lower()
    
    # Dictionary mapping month abbreviations and full names
    months = {
        'jan': 'January', 'feb': 'February', 'mar': 'March', 'apr': 'April',
        'may': 'May', 'jun': 'June', 'jul': 'July', 'aug': 'August',
        'sep': 'September', 'oct': 'October', 'nov': 'November', 'dec': 'December',
        'january': 'January', 'february': 'February', 'march': 'March', 
        'april': 'April', 'june': 'June', 'july': 'July', 'august': 'August', 
        'september': 'September', 'october': 'October', 'november': 'November', 
        'december': 'December'
    }
    
    # Check which month keyword is inside the string
    for key, month_name in months.items():
        if key in val_str:
            return month_name
            
    return None # Returns None if no valid month is found

# Apply it to create the new 'month' column
shark_attack_filtered.rename(columns={"date":"month"}, inplace=True)
shark_attack_filtered['month'] = shark_attack_filtered['month'].apply(extract_month)

print(shark_attack_filtered['month'])
```

```python
def clean_time_string(val):
    if pd.isna(val):
        return np.nan
    
    val_str = str(val).strip()
    
    # 1. Drop text descriptions, ranges, and noise
    if any(word in val_str.lower() for word in [
        'morning', 'afternoon', 'evening', 'night', 'dusk', 'midday', 
        'lunchtime', 'sunset', 'midnight', 'unknown', 'advised', 'stated', 
        'incident', 'before', 'after', 'between', 'just', 'shortly', '-', '/', 'or', '>'
    ]) or val_str in ['?', 'N', '--', 'Ê', '']:
        return np.nan

    # 2. Handle 12-hour format like '8:04 pm'
    if 'pm' in val_str.lower() or 'am' in val_str.lower():
        try:
            parsed = pd.to_datetime(val_str, format='%I:%M %p', errors='coerce')
            if not pd.isna(parsed):
                return parsed.strftime('%H:%M:00')
        except:
            pass

    # 3. Clean up and parse formats like '13h15' or '9h'
    cleaned = re.sub(r'[^0-9h]', '', val_str.lower())
    if 'h' in cleaned:
        parts = cleaned.split('h')
        hour = parts[0]
        minute = parts[1] if len(parts) > 1 and parts[1] != '' else '00'
        
        if hour.isdigit() and minute.isdigit():
            h_int, m_int = int(hour), int(minute)
            if 0 <= h_int < 24 and 0 <= m_int < 60:
                return f"{h_int:02d}:{m_int:02d}:00"
                
    # 4. Handle pure numeric strings (e.g., '1015', '1600', '800')
    digits_only = re.sub(r'\D', '', val_str)
    
    if len(digits_only) == 4:
        hour, minute = digits_only[:2], digits_only[2:]
    elif len(digits_only) == 3:
        hour, minute = '0' + digits_only[0], digits_only[1:]
    elif len(digits_only) == 2 or len(digits_only) == 1:
        hour, minute = digits_only.zfill(2), '00'
    else:
        return np.nan
        
    if hour.isdigit() and minute.isdigit():
        h_int, m_int = int(hour), int(minute)
        if 0 <= h_int < 24 and 0 <= m_int < 60:
            return f"{h_int:02d}:{m_int:02d}:00"
            
    return np.nan

# --- APPLY DIRECTLY TO YOUR EXISTING DATAFRAME ---

# 1. Apply the cleaning function to your column
shark_attack_final['cleaned_time'] = shark_attack_final['time'].apply(clean_time_string)

# 2. Convert to datetime.time datatype
shark_attack_final['cleaned_time'] = pd.to_datetime(shark_attack_final['cleaned_time'], format='%H:%M:%S', errors='coerce').dt.time

# 3. Find the mode and fill NaN values
mode_time = shark_attack_final['cleaned_time'].mode()[0] if not shark_attack_final['cleaned_time'].mode().empty else pd.to_datetime('00:00:00').time()
shark_attack_final['cleaned_time'] = shark_attack_final['cleaned_time'].fillna(mode_time)
```

```python
def clean_age(val):
    if pd.isna(val):
        return np.nan
        
    val_str = str(val).strip()
    
    # 1. Handle pure garbage/noise values
    if val_str in ['', '?', 'Ê', 'N', 'M', '6_']:
        return np.nan
        
    # 2. Handle descriptive text with no usable numbers
    val_lower = val_str.lower()
    if any(word in val_lower for word in ['middle age', 'teen']):
        return np.nan
        
    # 3. Handle months (e.g., '18 months' -> convert to 0 or round years)
    if 'month' in val_lower:
        digits = re.findall(r'\d+', val_str)
        return 0 if digits else np.nan

    # 4. Extract all digits from the string
    digits = re.findall(r'\d+', val_str)
    
    if not digits:
        return np.nan
        
    # 5. If it's a single number, decade ('50s', '50+'), or plus sign ('30+')
    if len(digits) == 1:
        return int(digits[0])
        
    # 6. If there are multiple numbers (ranges like '28 & 22', '45 and 15', '18 or 20')
    # We take the first number (or you can change this to mean/average if preferred)
    return int(digits[0])

# --- APPLY TO YOUR DATAFRAME ---

# 1. Apply the cleaning function to create a temporary cleaned series
shark_attack_final['cleaned_age'] = shark_attack_final['age'].apply(clean_age)

# 2. Fill missing (NaN) values with the mode (most frequent age)
mode_age = shark_attack_final['cleaned_age'].mode()[0] if not shark_attack_final['cleaned_age'].mode().empty else 30
shark_attack_final['cleaned_age'] = shark_attack_final['cleaned_age'].fillna(mode_age)

# 3. Assign the final result to your existing DataFrame as a new column
shark_attack_final['cleaned_age'] = shark_attack_final['cleaned_age'].astype(int)

# Verify the result
print(shark_attack_final[['age', 'cleaned_age']].head(20))
```
